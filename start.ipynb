```r
m <- glm(
  data = d,
  formula = Survived ~ Pclass + Sex + Pclass * Sex + Age + SibSp,
  family = binomial()
)

summary(m)
```

```r
# with factor.
m <- glm(
  data = d,
  formula = Survived ~ factor(Pclass) + Sex + factor(Pclass) * Sex + Age + SibSp,
  family = binomial()
)

summary(m)
```

```r
set.seed(100)
```

```r
d <- d %>%
  mutate(
    AgeGroup = case_when(
      Age < 20 ~ "under 20",
      Age > 50 ~ "over 50",
      TRUE     ~ "20-50"          # 20 to 50 inclusive
    ),
    AgeGroup = factor(AgeGroup, levels = c("under 20", "20-50", "over 50"))
  )

table(d$AgeGroup, useNA = "ifany")
```

```r
m <- glm(
  data = d,
  formula = Survived ~ factor(Pclass) + AgeGroup + factor(Pclass) * Sex + Age + SibSp,
  family = binomial()
)

summary(m)
```

```r
#type = "response" daje to zę wynikowy wektor jest typu 0,1 czyli zastosowany do sigmoidy

predictScore = m |> predict(newdata =  as_data_frame(titanic_test), type = "response")
predictScore
```
